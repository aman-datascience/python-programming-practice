# 1.Factorial of a number: 
Write a recursive function to calculate the factorial of a number n (i.e., n!)


```python
def factorial(n):
    if n==0 or n==1:
        return n
    return n*factorial(n-1)
n=int(input())
factorial(n)
```

# 2. Fibonacci series: 
Write a recursive function to generate the n-th number  of the Fibonacci sequence.

```python
def fibonacci(n):
    if n==1 or n==2:
        return 1
    else:
         return fibonacci(n-1)+fibonacci(n-2)
        
n=int(input('Enter the number:'))       
for i in range (1,n+1):
   print( fibonacci(i))
```

# 3. Sum of first n natural numbers:
Write a recursive function to calculate 
the sum of the first n natural numbers.

```python
def sum(n):
    if n==1:
        return 1
    return n+sum(n-1)
n=int(input('Enter the number:'))
sum(n)
```

## 4. Power of a number:
Write a recursive function to calculate x^n (x raised 
to the power of n).

```python
def power(x,n):
    if n==1:
        return x
    return x*power(x,n-1)
x=int(input('Enter the base:'))
n=int(input('Enter the power'))
power(x,n)
```

# 5. Reverse a string:
Write a recursive function to reverse a string.

```python
def reverse(s):
    if len(s)==0:
        return "" 
    return reverse(s[1:])+s[0]
reverse("Karan")
```

# 6. Sum of digits of a number:
Write a recursive function to calculate the 
sum of digits of a given number.

```python
def sod(n):
    if n<=0:
        return 0
    return sod(n//10)+n%10
n=int(input("Enter a number:"))
sod(n)
```

# 7. Count occurrences of a character in a string:
Write a recursive function 
to count how many times a particular character appears in a string.

```python
def count_character(s,c,count):
    if len(s)==0:
        return count
    elif s[len(s)-1]==c:
        count+=1
        return count_character(s[:len(s)-1:],c,count)
        
    else:
        return count_character(s[:len(s)-1:],c,count)
        
count_character('Aaaamaan','a',0)
        
```

# 8. Greatest Common Divisor (GCD):
Write a recursive function to compute 
the GCD of two numbers using Euclid’s algorithm

```python
def hcf(a,b):
    if a%b==0:
        return b
    return hcf(b,a%b)
hcf(88,33)
   
```

# 9. Check if a number is prime: 
Write a recursive function to check if a 
number is prime.

```python
def Prime_Number(n, i=2):
    if n == i:
        return True
    elif n % i == 0:
        return False
    return Prime_Number(n, i + 1)


n = 2
if Prime_Number(n):
    print("Yes,", n, "is Prime")
else:
    print("No,", n, "is not a Prime")
```

# 12.Find the minimum element in an array: 
Write a recursive function to  find the smallest element in an array.  

```python
def minimum(lst):
    if len(lst) == 1:
        return lst[0]

    x = minimum(lst[1:])

    if lst[0] < x:
        return lst[0]
    else:
        return x

lst = [5, 2, 8, 1, 4]
print(minimum(lst))
```

# 13 .Find the maximum element in an array:
Write a recursive function to find the largest element in an array.


```python
def minimum(lst):
    if len(lst) == 1:
        return lst[0]

    x = minimum(lst[1:])

    if lst[0] > x:
        return lst[0]
    else:
        return x

lst = [15, 2, 8, 1, 9]
print(minimum(lst))
```

# 16.Find the n-th triangular number:

# 20.Find the nth largest element in an array:
Write a recursive function to find the n-th largest element in an array


```python
def nth_largest(lst, n):
    if n == 1:
        return max(lst)

    x = max(lst)
    lst.remove(x)

    return nth_largest(lst, n - 1)

lst = [5, 2, 8, 1, 4]
print(nth_largest(lst, 2))
```
